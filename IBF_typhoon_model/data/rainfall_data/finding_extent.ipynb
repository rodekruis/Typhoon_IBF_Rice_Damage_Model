```python
import geopandas as gpd
import os
os.chdir("/home/jovyan/work/Typhoon_IBF_Rice_Damage_Model/")
cdir = os.getcwd()
# Load the shapefiles
typhoon_extents_path = 'IBF_typhoon_model/data/gis_data/typhoon_extents_tracks_combined/typhoon_extents_combined.shp'
admin_boundaries_path = 'IBF_typhoon_model/data/phl_administrative_boundaries/phl_admbnda_adm3.shp'

typhoon_extents = gpd.read_file(typhoon_extents_path)
admin_boundaries = gpd.read_file(admin_boundaries_path)

# Exclude this typhoon since it has an error
typhoon_extents = typhoon_extents[typhoon_extents['SID'] != '2010240N15142']

# Get the bounding boxes of both shapefiles
typhoon_bounds = typhoon_extents.total_bounds  # [minx, miny, maxx, maxy]
admin_bounds = admin_boundaries.total_bounds  # [minx, miny, maxx, maxy]

# Merge the bounding boxes to find the combined extent
combined_bounds = [
    min(typhoon_bounds[0], admin_bounds[0]),  # minx
    min(typhoon_bounds[1], admin_bounds[1]),  # miny
    max(typhoon_bounds[2], admin_bounds[2]),  # maxx
    max(typhoon_bounds[3], admin_bounds[3])   # maxy
]

# Print the combined bounding box
print("Combined Bounding Box:")
print(f"Min Longitude: {combined_bounds[0]}")
print(f"Min Latitude: {combined_bounds[1]}")
print(f"Max Longitude: {combined_bounds[2]}")
print(f"Max Latitude: {combined_bounds[3]}")

# Print the bounding box in the required format
print("GES DISC Bounding Box Format:")
print(f"{combined_bounds[0]}, {combined_bounds[1]}, {combined_bounds[2]}, {combined_bounds[3]}")
```
